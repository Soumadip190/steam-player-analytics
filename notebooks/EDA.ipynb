```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("steam-200k.csv")
df.head()
```

```python
# Remove redundant column
df_new = df.copy()

df.columns = [
    "user_id",
    "game_title",
    "behavior",
    "value",
    "extra_column"
]

# Remove constant column
df = df.drop(columns=["extra_column"])

# Remove exact duplicates
df = df.drop_duplicates().copy()

# Separate behaviour
purchase_df = df[df["behavior"] == "purchase"].copy()
play_df = df[df["behavior"] == "play"].copy()

# Player-level features
player_features = df.groupby("user_id").agg(
    total_interactions=("game_title", "count"),
    unique_games=("game_title", "nunique")
)

# Total gameplay hours
total_hours = play_df.groupby("user_id")["value"].sum()

# Unique purchased games
total_purchases = purchase_df.groupby("user_id")["game_title"].nunique()

player_features["total_hours"] = total_hours
player_features["total_purchases"] = total_purchases

player_features = player_features.fillna(0)

display(player_features.head(10))

print("Player dataset shape:", player_features.shape)
```
